```python
import pyomo.environ as pyo

m = pyo.ConcreteModel(name="starterMwuhahaha")

m.x = pyo.Var(domain=pyo.NonNegativeIntegers)
m.y = pyo.Var(domain=pyo.NonNegativeIntegers)

m.profit = pyo.Objective(expr=m.x + 2 * m.y, sense=pyo.minimize)

m.one = pyo.Constraint(expr=m.x + m.y >= 10)
m.two = pyo.Constraint(expr=2 * m.x + m.y >= 5)
m.three = pyo.Constraint(expr=m.y >= 3)

solver = pyo.SolverFactory("gurobi")
# results = solver.solve(m, tee=True)
results = solver.solve(m)

print("Status:", results.solver.termination_condition)
print(f"x = {pyo.value(m.x)}")
print(f"y = {pyo.value(m.y)}")
print(f"Objective = {pyo.value(m.profit)}")
```
